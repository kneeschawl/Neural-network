LAB 6: Unsupervised Learning

OBJECTIVE:  Perform clustering using K-Means from scratch.

DATASET: Iris Dataset

THEORY:

- Unsupervised learning finds patterns in data without labels.

- K-Means clustering partitions data into k clusters by:

    - Randomly initializing k centroids

    - Assigning each point to the nearest centroid

    - Updating centroids based on the mean of points in each cluster

    - Repeating until convergence

- Inertia measures how close data points are to their assigned cluster centers (lower is better).

- Tools like Scikit-learn simplify clustering and visualization.

ALGORITHM (K-MEANS CLUSTERING):

- Load dataset (Iris or synthetic)

- Select the number of clusters k

- Initialize centroids randomly

- Repeat until convergence:

    - Assign each data point to the nearest centroid

    - Update centroids as the mean of assigned points

- Compute inertia (sum of squared distances)

- Visualize clusters and centroids

SOURCE CODE:

CELL 1: IMPORT LIBRARIES

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

CELL 2: LOAD DATASET

```python
# Load Iris dataset
iris = load_iris()
X = iris.data[:, :2]  # use first two features for easy 2D visualization

print("Features shape:", X.shape)
```

CELL 3: APPLY K-MEANS

```python
# Define number of clusters
k = 3

# Initialize and fit KMeans
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(X)

# Get cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# Inertia
print("Inertia (sum of squared distances):", kmeans.inertia_)
```

CELL 4: VISUALIZE CLUSTERS

```python
plt.figure(figsize=(8,6))

# Plot clusters
for i in range(k):
    plt.scatter(X[labels==i, 0], X[labels==i, 1], label=f'Cluster {i}')

# Plot centroids
plt.scatter(centroids[:,0], centroids[
    :,1], s=200, c='red', marker='X', label='Centroids')

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clustering on Iris Dataset")
plt.legend()
plt.show()
```

RESULTS:

- K-Means clustered the Iris dataset into 3 distinct groups, mostly aligning with actual species.

- Cluster centroids are clearly visible on the plot.

- Inertia (sum of squared distances) was low, indicating compact clusters.

- Visualization confirms that similar points are grouped together effectively.

CONCLUSION:

- K-Means is effective for unsupervised clustering when the number of clusters is known.

- The algorithm groups similar data points without using labels.

- Inertia provides a quantitative measure of cluster compactness, while visualization helps interpret the clusters.

- This lab demonstrates the ability of unsupervised learning to uncover patterns in unlabeled data.
